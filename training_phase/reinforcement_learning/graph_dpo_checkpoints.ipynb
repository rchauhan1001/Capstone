```python
import json
import matplotlib.pyplot as plt
import numpy as np

DATA_DIR = "/Users/eitan/Documents/School related/Capstone Project/training_phase/reinforcement_learning"

with open(f"{DATA_DIR}/eval_history_beta1.0_lr5e-06.json") as f:
    hist_beta1 = json.load(f)

with open(f"{DATA_DIR}/eval_history_beta2.0_lr5e-06.json") as f:
    hist_beta2 = json.load(f)

def extract(history, key):
    return [e["step"] for e in history], [e[key] for e in history]
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

steps1, margins1 = extract(hist_beta1, "eval_rewards/margins")
steps2, margins2 = extract(hist_beta2, "eval_rewards/margins")

ax.plot(steps1, margins1, "o-", label="beta=1.0", color="#e74c3c")
ax.plot(steps2, margins2, "s-", label="beta=2.0", color="#2980b9")
ax.axhspan(1.0, 2.0, alpha=0.15, color="green", label="Target range (1.0-2.0)")
ax.set_xlabel("Training Step")
ax.set_ylabel("Reward Margin (chosen - rejected)")
ax.set_title("DPO Reward Margins Over Training")
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(f"{DATA_DIR}/margins_over_training.png", dpi=150)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

steps1, acc1 = extract(hist_beta1, "eval_rewards/accuracies")
steps2, acc2 = extract(hist_beta2, "eval_rewards/accuracies")

ax.plot(steps1, acc1, "o-", label="beta=1.0", color="#e74c3c")
ax.plot(steps2, acc2, "s-", label="beta=2.0", color="#2980b9")
ax.axhline(y=0.90, linestyle="--", color="gray", alpha=0.5, label="90% threshold")
ax.set_xlabel("Training Step")
ax.set_ylabel("Reward Accuracy")
ax.set_title("DPO Reward Accuracy Over Training")
ax.set_ylim(0.55, 1.02)
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(f"{DATA_DIR}/accuracy_over_training.png", dpi=150)
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)

for ax, hist, beta, color in [(axes[0], hist_beta1, "1.0", "#e74c3c"),
                                (axes[1], hist_beta2, "2.0", "#2980b9")]:
    steps_c, logp_c = extract(hist, "eval_logps/chosen")
    steps_r, logp_r = extract(hist, "eval_logps/rejected")

    ax.plot(steps_c, logp_c, "o-", label="Chosen", color="#27ae60")
    ax.plot(steps_r, logp_r, "s-", label="Rejected", color="#c0392b")
    ax.set_xlabel("Training Step")
    ax.set_ylabel("Log Probability")
    ax.set_title(f"Log-Probs: beta={beta}")
    ax.legend()
    ax.grid(True, alpha=0.3)

plt.suptitle("Chosen vs Rejected Log-Probabilities Over Training", fontsize=13)
plt.tight_layout()
plt.savefig(f"{DATA_DIR}/logprobs_over_training.png", dpi=150)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

steps1, loss1 = extract(hist_beta1, "eval_loss")
steps2, loss2 = extract(hist_beta2, "eval_loss")

ax.plot(steps1, loss1, "o-", label="beta=1.0", color="#e74c3c")
ax.plot(steps2, loss2, "s-", label="beta=2.0", color="#2980b9")
ax.set_xlabel("Training Step")
ax.set_ylabel("Eval Loss")
ax.set_title("DPO Eval Loss Over Training")
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(f"{DATA_DIR}/eval_loss_over_training.png", dpi=150)
plt.show()
```
