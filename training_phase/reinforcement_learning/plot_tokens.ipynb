```python
import matplotlib.pyplot as plt
import numpy as np

labels = ['Min', 'Mean', 'Median', 'Max']
chosen =   [94, 161, 159, 361]
rejected = [79, 122, 120, 352]

x = np.arange(len(labels))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 5))
bars1 = ax.bar(x - width/2, chosen, width, label='Chosen', color='#27ae60', alpha=0.8)
bars2 = ax.bar(x + width/2, rejected, width, label='Rejected', color='#c0392b', alpha=0.8)

ax.set_xlabel('Statistic')
ax.set_ylabel('Token Length (prompt + completion)')
ax.set_title('Token Length Distribution: Chosen vs Rejected')
ax.set_xticks(x)
ax.set_xticklabels(labels)
ax.legend()
ax.grid(True, alpha=0.3, axis='y')

for bar in bars1:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 5,
            f'{int(bar.get_height())}', ha='center', va='bottom', fontsize=9)
for bar in bars2:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 5,
            f'{int(bar.get_height())}', ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.savefig('token_length_distribution.png', dpi=150)
plt.show()
```
